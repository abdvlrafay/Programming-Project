Setup and Constant: <br>
import csv is a built in module for reading and writing comma seperated value files<br>
line 2 defines a global constant holding the file name <br>

```python
import csv
EVENT_FILE = "uni_events.csv"
```

4 Pillars of Object Oriented Programming: <br>
Abstraction: BasePortal (line3) acts as an outline or interface of what operations should exist without implementing immediately <br>
Inheritance: line 4 inherits the structure from parent BasePortal Class <br>
Encapsulation: (self.__filepath) Restricts external access to private variables inside class <br>
Polymorphism: (display_report) overrides the parent class display report to provide specific csv reading <br>

```python
class BasePortal:
    def display_report(self):
        pass
        
class EventPortal(BasePortal):
    def __init__(self, file_path):
        self.__file_path = file_path
        
    def display_report(self):
        try:
            print("\n=== University Events Report ===")
            with open(self.__file_path, mode="r") as file:
                csv_reader = csv.reader(file)
                next(csv_reader, None)
                for event_row in csv_reader:
                    if event_row:
                        print(f"Event: {event_row[0]} | Date: {event_row[1]} | Seats: {event_row[2]} | Ticket Price: {event_row[3]}")
        except FileNotFoundError:
            print("No events found yet.")
```

Core Functions (Login, adding event record, searching, and generating a report): <br>
login() handles admin credentials check <br>
add_record() collects event inputs and checks if the file is new or already existing to write table headers or append the new row <br>
search_record() skims the csv file row by row (skipping headers) to display the searched events details <br>
generate_report() creates an object from EventPortal class to display the full list of saved events in csv file <br>

```python
def login():
    username = input("Enter your username: ")
    password = input("Enter your password: ")
    return username == "admin" and password == "1234"

def add_record():
    try:
        event_name = input("Enter Event Name: ")
        event_date = input("Enter Date: ")
        seat_capacity = int(input("Enter Seats: "))
        tick_price= int(input("Enter ticket price: "))
        try:
            with open(EVENT_FILE, "r") as file:
                is_file_empty = len(file.read()) == 0
        except FileNotFoundError:
            is_file_empty = True
        with open(EVENT_FILE, mode="a", newline="") as file:
            csv_writer = csv.writer(file)
            if is_file_empty:
                csv_writer.writerow(["Event Name", "Date", "Seats", "Ticket Price"])
            csv_writer.writerow([event_name, event_date, seat_capacity, tick_price])
        print("Event added successfully")
    except ValueError:
        print("Error, Seats and ticket price both must be an integer type")

def search_record():
    search = input("Enter event name to search: ")
    try:
        with open(EVENT_FILE, mode="r") as file:
            csv_reader = csv.reader(file)
            next(csv_reader, None) 
            is_found = False
            for event_row in csv_reader:
                if event_row and event_row[0].lower() == search.lower():
                    print("\nEvent Found")
                    print(f"Name: {event_row[0]}")
                    print(f"Date: {event_row[1]}")
                    print(f"Seats: {event_row[2]}")
                    print(f"Ticket Price: {event_row[3]}")
                    is_found = True
            if not is_found:
                print("Event not found")
    except FileNotFoundError:
        print("Event file does not exists yet")

def generate_report():
    event_portal = EventPortal(EVENT_FILE)
    event_portal.display_report()
```

Main Program: <br>
Execeutes the system workflow <br>
First prompts the user for login via login() <br>
If the authentication check is passed, it moves onto interactive while loop which keeps the portal choices active until user chooses option 4 to exit the portal. <br>

```python
if login():
    print("\nLogined Successfully")
    while True:
        print("\n===== UNIVERSITY EVENT MANAGEMENT PORTAL =====")
        print("1.) Add a new event")
        print("2.) Search an event")
        print("3.) Generate events report")
        print("4.) Exit the portal")

        choice = input("Choose an option: ")
        if choice == '1': add_record()
        elif choice == '2': search_record()
        elif choice == '3': generate_report()
        elif choice == '4':
            print("Thank you for using the Event Management System!")
            break
        else:
            print("Invalid option, please try again")
else:
    print("Incorrect username or password entered")
```
